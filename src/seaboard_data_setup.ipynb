```python
import open3d as o3d
import plotly.graph_objects as go
import numpy as np
import os

# Base folder path for the Porites Dataset (November 2025)
base_path = '/content/drive/MyDrive/coral seaboard data/Porites Dataset (November 2025)/'

# All seaboards with their folder names and corresponding .obj file names
seaboards = [
    {"folder": "Seaboard 1 (Stefan)",  "obj": "seaboard_Stefan_1.obj",  "label": "Seaboard 1 (Stefan)"},
    {"folder": "Seaboard 1 (Taitusi)", "obj": "seaboard_Taitusi_1.obj", "label": "Seaboard 1 (Taitusi)"},
    {"folder": "Seaboard 2 (Stefan)",  "obj": "seaboard_Stefan_2.obj",  "label": "Seaboard 2 (Stefan)"},
    {"folder": "Seaboard 2 (Taitusi)", "obj": "seaboard_Taitusi_2.obj", "label": "Seaboard 2 (Taitusi)"},
    {"folder": "Seaboard 3",           "obj": "seaboard_Taitusi_3.obj", "label": "Seaboard 3"},
    {"folder": "Seaboard 4",           "obj": "seaboard_4.obj",         "label": "Seaboard 4"},
    {"folder": "Seaboard 5",           "obj": "seaboard_5.obj",         "label": "Seaboard 5"},
    {"folder": "Seaboard 6",           "obj": "seaboard_6.obj",         "label": "Seaboard 6"},
    {"folder": "Seaboard 7",           "obj": "seaboard_7.obj",         "label": "Seaboard 7"},
    {"folder": "Seaboard 8",           "obj": "seaboard_8.obj",         "label": "Seaboard 8"},
    {"folder": "Seaboard 9",           "obj": "seaboard_9.obj",         "label": "Seaboard 9"},
    {"folder": "Seaboard 10",          "obj": "seaboard_10.obj",        "label": "Seaboard 10"},
]

def create_obj_plot(file_path, title):
    """Load a .obj file with open3d and return a Plotly figure."""
    if not os.path.exists(file_path):
        print(f"File not found: {file_path}")
        return None

    mesh = o3d.io.read_triangle_mesh(file_path)
    mesh.compute_vertex_normals()

    verts  = np.asarray(mesh.vertices)
    faces  = np.asarray(mesh.triangles)
    has_colors = mesh.has_vertex_colors()
    colors = np.asarray(mesh.vertex_colors) if has_colors else None

    if has_colors:
        # Convert RGB floats [0,1] to hex strings for vertexcolor
        colors_uint8 = (colors * 255).astype(np.uint8)
        vertex_colors = [f'rgb({r},{g},{b})' for r, g, b in colors_uint8]
        trace = go.Mesh3d(
            x=verts[:, 0], y=verts[:, 1], z=verts[:, 2],
            i=faces[:, 0], j=faces[:, 1], k=faces[:, 2],
            vertexcolor=vertex_colors,
            flatshading=False,
            lighting=dict(ambient=0.4, diffuse=0.8, specular=0.2),
        )
    else:
        # No texture — use Z-height as intensity for a visible colorscale
        intensity = verts[:, 2]
        trace = go.Mesh3d(
            x=verts[:, 0], y=verts[:, 1], z=verts[:, 2],
            i=faces[:, 0], j=faces[:, 1], k=faces[:, 2],
            intensity=intensity,
            colorscale='Viridis',
            showscale=False,
            flatshading=True,
            lighting=dict(ambient=0.4, diffuse=0.8, specular=0.2),
        )

    fig = go.Figure(trace)
    fig.update_layout(
        title=title,
        scene=dict(
            aspectmode='data',
            xaxis_visible=False,
            yaxis_visible=False,
            zaxis_visible=False,
        ),
        margin=dict(l=0, r=0, b=0, t=40)
    )
    return fig
```

```python
sb = seaboards[0]
obj_path = os.path.join(base_path, sb["folder"], sb["obj"])
print(f"Loading: {obj_path}")
fig = create_obj_plot(obj_path, sb["label"])
if fig: fig.show()
```
